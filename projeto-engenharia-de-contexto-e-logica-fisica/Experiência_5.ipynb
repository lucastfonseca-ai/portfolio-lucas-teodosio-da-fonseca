```python
# Algoritmo 1: Microclima Local

def analisar_microclima():
  print("--- Analisador de Microclima Urbano ---")
  print("--- Trajeto diário: Casa → Unicid (Itaquera) ---")

  # Lista de Listas com os dados coletados de cada local
  # Estrutura: [Nome, Temp_Manhã, Umidade_Manhã, IQA_Manhã, Temp_Noite, Umidade_Noite, IQA_Noite]
  # IQA numérico: Boa=20, Moderada=60, Ruim=100, Muito Ruim=160, Péssima=220

  dados_locais = [
    ["Ponto de Ônibus (Sem Cobertura)", 22, 0, 60, 20, 23, 100],
    ["Metrô Itaquera", 24, 0, 20, 21, 9, 60],
    ["Terminal de Ônibus (Carrão)", 20, 10, 60, 19, 14, 60],
  ]

  # Nomes dos períodos para exibição
  periodos = ["Manhã", "Noite"]

  for local in dados_locais:

    # Desestrutura o nome e as métricas do local atual
    nome = local[0]
    print(f"\n{'=' * 58}")
    print(f"Local: {nome}")

    # Agrupa as métricas por período:
    # índices [1, 2, 3] = manhã | índices [4, 5, 6] = noite
    metricas_por_periodo = [
      [local[1], local[2], local[3]],
      [local[4], local[5], local[6]],
    ]

    # Loop aninhado: itera sobre cada período do local atual
    for i, metricas in enumerate(metricas_por_periodo):

      temperatura = metricas[0] # Em graus Celsius
      umidade = metricas[1] # Em porcentagem (%)
      iqa = metricas[2] # Índice de Qualidade do Ar (numérico oficial)
      periodo = periodos[i]

      print(f"\n Periodo: {periodo}")
      print(f"Temperatura: {temperatura} C")
      print(f"Umidade: {umidade}%")
      print(f"IQA: {iqa}")

      # match-case para classificar o IQA com base nas faixas oficiais do IBAMA
      # Usa um "truque": avalia expressões booleanas dentro do match
      match True:

        case _ if iqa <= 40:
          classificacao_iqa = "Boa"
          penalidade_iqa  = 0.0
          print(f"Qualidade do Ar: Boa (IQA de 0 a 40 - Sem riscos a saude)")

        case _ if iqa <= 80:
          classificacao_iqa = "Moderada"
          penalidade_iqa  = 2.0
          print(f"Qualidade do Ar: Moderada (IQA de 41 a 80 - Grupos sensiveis atentos)")

        case _ if iqa <= 120:
          classificacao_iqa = "Ruim"
          penalidade_iqa  = 4.5
          print(f"Qualidade do Ar: Ruim (IQA de 81 a 120 - Toda populacao afetada)")

        case _ if iqa <= 200:
          classificacao_iqa = "Muito Ruim"
          penalidade_iqa  = 7.0
          print(f"Qualidade do Ar: Muito Ruim (IQA de 121 a 200 - Falta de ar e risco grave)")

        case _:
          classificacao_iqa = "Pessima"
          penalidade_iqa  = 10.0
          print(f"Qualidade do Ar: Pessima (IQA acima de 200 - Emergencia respiratoria)")

      # --- Formula da Nota de Conforto Urbano (0 a 10) ---
      # Temperatura humana ideal: entre 18 C e 26 C (OMS/INMET)
      # Umidade ideal: entre 40% e 70% (ANVISA)
      # Penaliza desvios fora dessas faixas + penalidade do IQA
      # Penalidade de temperatura: quanto mais longe do intervalo ideal, maior a penalidade
      temp_ideal_min = 18
      temp_ideal_max = 26
      if temp_ideal_min <= temperatura <= temp_ideal_max:
        penalidade_temp = 0.0
      elif temperatura < temp_ideal_min:

        # Cada grau abaixo do mínimo pesa 0.4 pontos
        penalidade_temp = (temp_ideal_min - temperatura) * 0.4
      else:

        # Cada grau acima do máximo pesa 0.6 pontos (calor é mais prejudicial)
        penalidade_temp = (temperatura - temp_ideal_max) * 0.6

      # Penalidade de umidade: extremos (muito seco ou muito úmido) são prejudiciais
      umidade_ideal_min = 40
      umidade_ideal_max = 70
      if umidade_ideal_min <= umidade <= umidade_ideal_max:
        penalidade_umidade = 0.0
      elif umidade < umidade_ideal_min:

        # Ar muito seco: resseca mucosas e vias respiratórias
        penalidade_umidade = (umidade_ideal_min - umidade) * 0.06
      else:

        # Ar muito úmido: favorece fungos e sensação de abafamento
        penalidade_umidade = (umidade - umidade_ideal_max) * 0.04

      # Calcula a nota final cruzando temperatura, umidade e IQA
      nota_conforto = 10.0 - penalidade_temp - penalidade_umidade - penalidade_iqa

      # Operador lógico: garante que a nota fique sempre entre 0 e 10
      nota_conforto = max(0.0, min(10.0, nota_conforto))
      print(f"\n Nota de Conforto Urbano: {nota_conforto:.1f} / 10")

      # Interpretação da nota com operadores lógicos compostos
      if nota_conforto >= 7.0 and classificacao_iqa in ["Boa", "Moderada"]:
        print(f"Condicao: CONFORTAVEL - Bom para deslocamento")
      elif nota_conforto >= 4.0 or classificacao_iqa == "Moderada":
        print(f"Condicao: TOLERAVEL - Atencao recomendada")
      elif nota_conforto >= 2.0 and classificacao_iqa in ["Ruim", "Muito Ruim"]:
        print(f"Condicao: DESCONFORTAVEL - Evite exposicao prolongada")
      else:
        print(f"Condicao: CRITICA - Risco a saude. Use mascara!")

  print(f"\n{'=' * 58}")
  print("Analise de microclima concluida.")
  print(f"{'=' * 58}")

analisar_microclima()
```

```python
# Algoritmo 2: Navegação e Evacuação Espacial

import random
def simular_evacuacao():
  print("--- Simulador de Evacuacao de Incendio ---")
  print("--- Bloco C, 1 Andar - Unicid ---")

  # Lista 1: Nomes dos nos (locais/corredores) do trajeto de evacuacao
  # O agente tenta chegar do indice 0 ate o ultimo indice (saida)
  locais = [
    "Sala de Aula (Bloco C)", # 1 - Ponto de partida
    "Corredor Principal",
    "Escadas de Emergencia (Lateral)", # 2 - Saida mais proxima ao Bloco C
    "Saida Lateral - Rua", # 3 - Objetivo final
  ]

  # Lista 2: Estado de cada no (obstaculos ou facilitadores mapeados)
  # Correspondente por indice a lista de locais
  estados = [
    "fumaca densa", # 0 - Fumaca tomando a sala
    "multidao desesperada", # 1 - Corredor lotado de alunos em panico
    "livre", # 2 - Escadas de emergencia desobstruidas
    "saida", # 3 - Rua / Evacuado com sucesso
  ]

  # Inventario do agente: itens que ele carrega e que desbloqueiam acoes especiais
  # O agente pode encontrar itens ao longo do trajeto
  inventario = ["toalha_umida"] # Começa apenas com uma toalha umida

  # Variaveis de controle do agente
  posicao_atual = 0   # Indice do local atual na lista
  energia = 100  # Energia do agente (chega a 0 = falha na evacuacao)
  agente_saiu = False # Flag de controle: True quando o agente alcanca a saida
  tentativas = 0   # Contador de tentativas do loop while
  max_tentativas = 25  # Limite de seguranca para evitar loop infinito

  print(f"\nAgente iniciando evacuacao | Energia: {energia} | Inventario: {inventario}\n")

  # Loop while: o agente continua tentando enquanto nao saiu, tem energia e nao esgotou tentativas
  while not agente_saiu and energia > 0 and tentativas < max_tentativas:

    tentativas += 1
    local_atual = locais[posicao_atual]
    obstaculo = estados[posicao_atual]

    print(f"[Tentativa {tentativas:02}] Local atual: '{local_atual}'")
    print(f"Obstaculo: '{obstaculo}' | Energia: {energia}")

    # ---- Avaliacao do no atual com if aninhado ----
    if obstaculo == "saida":

      # Condicao de vitoria: agente alcancou a rua
      print("SAIDA ENCONTRADA! Agente evacuado com sucesso!\n")
      agente_saiu = True
    elif obstaculo == "fumaca densa":
      print("Fumaca densa detectada!")

      # Sub-condicao 1: tem toalha umida no inventario?
      if "toalha_umida" in inventario:
        print("-> Agente usa a toalha umida para filtrar o ar. Avancando...")
        inventario.remove("toalha_umida") # Consome o item
        energia -= 15
        posicao_atual += 1 # Avanca para o proximo no

      # Sub-condicao 2: nao tem toalha, mas tem extintor para abrir caminho?
      elif "extintor" in inventario:
        print("-> Sem toalha, mas usa o extintor para dispersar a fumaca. Avancando com risco...")
        inventario.remove("extintor")
        energia -= 25
        posicao_atual += 1

      # Sub-condicao 3: nao tem nenhum item. Arrisca atravessar na forca bruta
      else:
        print("-> Sem itens! Agente avanca pela fumaca sem protecao...")
        energia -= 40 # Penalidade alta por exposicao direta
        if energia > 0:
          posicao_atual += 1
        else:
          print("CRITICO: Agente desmaiou por intoxicacao por fumaca!")
    elif obstaculo == "multidao desesperada":
      print("Multidao em panico bloqueando o corredor!")

      # Sub-condicao 1: brigada de incendio presente no inventario (como aliado)?
      if "brigada" in inventario:
        print("-> Brigada de incendio coordena a multidao. Corredor aberto!")
        energia -= 10
        posicao_atual += 1

      # Sub-condicao 2: nao tem brigada, o agente decide aleatoriedade do caos
      else:

        # Matematica: simula o caos com aleatoriedade ponderada
        chance = random.randint(1, 100)

        if chance <= 50:

          # 50% de chance de conseguir passar empurrando
          print("-> Agente empurra com determinacao e passa pela multidao. Avancando...")
          energia -= 25
          posicao_atual += 1
        elif chance <= 80:

          # 30% de chance de ficar preso e perder mais energia esperando
          print("-> Agente fica preso na multidao por alguns segundos. Perdendo energia...")
          energia -= 20
          # Nao avanca nem retrocede, tenta de novo na proxima iteracao
        else:

          # 20% de chance de retroceder para buscar rota alternativa
          print("-> Agente recua e busca rota alternativa pelo corredor lateral...")
          energia -= 12
          posicao_atual = max(0, posicao_atual - 1) # Retrocede sem sair dos limites
    elif obstaculo == "destrocos":
      print("Destrocos bloqueando a passagem!")

      # Sub-condicao: tem energia suficiente para remover destrocos?
      if energia >= 30:
        print("-> Agente remove destrocos com esforco fisico. Avancando lentamente...")
        energia -= 25
        posicao_atual += 1
      else:
        print("-> Agente sem energia suficiente para remover destrocos. Retrocede...")
        energia -= 10
        posicao_atual = max(0, posicao_atual - 1)
    elif obstaculo == "porta trancada":
      print("Porta trancada detectada!")

      # Sub-condicao 1: tem chave no inventario?
      if "chave" in inventario:
        print("-> Agente usa a chave. Porta aberta. Avancando...")
        inventario.remove("chave")
        energia -= 5
        posicao_atual += 1

      # Sub-condicao 2: nao tem chave, mas pode arrombar com extintor?
      elif "extintor" in inventario:
        print("-> Sem chave. Agente usa o extintor para arrombar a porta. Avancando...")
        inventario.remove("extintor")
        energia -= 20
        posicao_atual += 1

      # Sub-condicao 3: nenhum item disponivel. Retrocede
      else:
        print("-> Sem chave e sem extintor. Agente retrocede para buscar alternativa...")
        energia -= 15
        posicao_atual = max(0, posicao_atual - 1)
    elif obstaculo == "livre":

      # No livre: sem obstaculos, avanca normalmente com custo minimo de energia
      print("Caminho livre! Agente avanca sem obstaculo.")
      energia -= 5
      posicao_atual += 1
    else:

      # Caso inesperado (obstaculos nao mapeados)
      print(f"Obstaculo desconhecido: '{obstaculo}'. Agente recalcula rota...")
      energia -= 10
      posicao_atual = max(0, posicao_atual - 1)

    # Operador logico: garante que a posicao nunca ultrapasse o limite da lista
    posicao_atual = min(posicao_atual, len(locais) - 1)
    print(f"Inventario: {inventario}\n")

  # --- Resultado final da simulacao ---
  print("=" * 50)

  if agente_saiu:
    print(f"EVACUACAO BEM-SUCEDIDA!")
    print(f"Tentativas utilizadas: {tentativas}")
    print(f"Energia restante: {max(0, energia)}")
  elif energia <= 0:
    print(f"EVACUACAO FALHOU: Agente sem energia no local '{locais[posicao_atual]}'.")
  else:
    print(f"EVACUACAO FALHOU: Limite de {max_tentativas} tentativas atingido.")

  print(f"Posicao final: '{locais[posicao_atual]}'")
  print("=" * 50)

simular_evacuacao()
```
