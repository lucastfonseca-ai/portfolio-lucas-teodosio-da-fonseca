```python
# Estrutura de Dados
import time
empresa_data = {
    "Google": {
        "Tecnologia": {
            "Cloud": {
                "Servidores": 80000,
                "IA": 120000
            },
            "Android": 60000
        },

        "YouTube": {
            "Creators": 30000,
            "Premium": 25000
        },

        "RH": {
            "Recrutamento": 15000,
            "Treinamento": 10000
        },

        "Marketing": 40000
    }
}

google = empresa_data["Google"]
print("Empresa:", google)
```

```python
# Decorator
def auditor(funcao_original):
    def wrapper(nome_da_empresa, *departamentos_ignorados, **parametros_financeiros):
        print("\n========== AUDITORIA FINANCEIRA ==========")
        print(f"Empresa Multinacional: {nome_da_empresa}")
        inicio = time.time()

        print("\nDepartamentos Ignorados (*args):")
        for departamento in departamentos_ignorados:
            print(f"- {departamento}")

        print("\nParâmetros Financeiros (**kwargs):")
        for chave, valor in parametros_financeiros.items():
            print(f"- {chave}: {valor}")

        resultado = funcao_original(
            nome_da_empresa,
            *departamentos_ignorados,
            **parametros_financeiros
        )
        fim = time.time()

        print(f"\nOrçamento Total Convertido: R$ {resultado:.2f}")
        print(f"Tempo de execução: {fim - inicio:.6f} segundos")
        print("==========================================")
        return resultado
    return wrapper

@auditor
def calcular_orcamento(nome_da_empresa, *departamentos_ignorados, **parametros_financeiros):
    taxa = parametros_financeiros.get("taxa_cambio", 1)

    def somar(dados):
        total = 0

        for chave, valor in dados.items():
            if chave in departamentos_ignorados:
                continue
            if isinstance(valor, dict):
                total += somar(valor)
            else:
                total += valor
        return total
    total_final = somar(empresa_data[nome_da_empresa])
    return total_final * taxa
```

```python
calcular_orcamento(

    # 1º Parâmetro
    "Google",

    # 2º Parâmetro (*args)
    "RH",
    "Marketing",

    # 3º Parâmetro (**kwargs)
    moeda_destino="BRL",
    taxa_cambio=5.20,
    auditoria="Ativada",
    responsavel="Equipe Financeira",
    ano_fiscal=2026
)
```
