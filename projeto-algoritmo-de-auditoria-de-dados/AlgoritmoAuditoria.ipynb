```python
LIMITE_SEGURANCA = 10000.00

# Valor das vendas
print("Digite o valor das vendas:")
venda1 = float(input ("Valor da Venda 1: "))
venda2 = float(input ("Valor da Venda 2: "))
venda3 = float(input ("Valor da Venda 3: "))

# Média das vendas
def analisar_vendas(venda1,venda2,venda3):
  global LIMITE_SEGURANCA # Permite alterar o limite dentro da função

  media = (venda1 + venda2 + venda3) / 3
  print(f"Média das vendas: R${media:.2f}")
  return media

media_das_vendas = analisar_vendas(venda1,venda2,venda3)

if media_das_vendas >= LIMITE_SEGURANCA:
  print("SISTEMA EM QUARENTENA")
else:
  print("MÉDIA NORMAL")

# Revisão Manual
if (
    venda1 > LIMITE_SEGURANCA or
    venda2 > LIMITE_SEGURANCA or
    venda3 > LIMITE_SEGURANCA
):
    resposta = input("A venda é legítima? (s/n): ").lower()

    if resposta == "s":
        print("Limite de segurança aumentado!")
        LIMITE_SEGURANCA *= 10 # Aumentar o limite
        print(f"Novo limite: {LIMITE_SEGURANCA}")
    elif resposta == "n":
        print("Limite de segurança mantido.")
else:
    print("Nenhuma venda ultrapassou o limite.")

# Tipo de variáveis
print("\n=== Tipos de Dados ===")
print(f"Variável da Venda 1: {type(venda1)}")
print(f"Variável da Venda 2: {type(venda2)}")
print(f"Variável da Venda 3: {type(venda3)}")
print(f"Variável da Média: {type(media_das_vendas)}")
print(f"Variável do Limite de Serugança: {type(LIMITE_SEGURANCA)}")
```
